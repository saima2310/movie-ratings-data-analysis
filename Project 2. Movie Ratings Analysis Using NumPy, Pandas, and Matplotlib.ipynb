```python
import pandas as pd

# Part 2: Data Cleaning

# Load the movie ratings dataset
df = pd.read_csv('movie_ratings_data.csv')

# 1. Check for missing (null) values in each column
print("--- Missing Values Before Cleaning ---")
print(df.isnull().sum())
```

```python
# 2. Handle missing values
# Fill numerical columns (Rating, Revenue_Million) with their respective mean values
df['Rating'] = df['Rating'].fillna(df['Rating'].mean())
df['Revenue_Million'] = df['Revenue_Million'].fillna(df['Revenue_Million'].mean())

# Verify if any null values remain
print("\n--- Missing Values After Cleaning ---")
print(df.isnull().sum())
```

```python
# 3. Check for duplicate rows and remove them
print(f"\nNumber of duplicate rows before removal: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"Number of duplicate rows after removal: {df.duplicated().sum()}")

# 4. Save the cleaned dataset back to CSV
df.to_csv('cleaned_movie_ratings_data.csv', index=False)
print("\nData cleaning completed and saved as 'cleaned_movie_ratings_data.csv'!")
```

```python
import pandas as pd
import numpy as np

# Part 3: Statistical Analysis with NumPy

# Load the cleaned movie ratings dataset
df = pd.read_csv('cleaned_movie_ratings_data.csv')

# 1. Rating Statistics
rating_mean = np.mean(df['Rating'])
rating_median = np.median(df['Rating'])
rating_std = np.std(df['Rating'])

print("--- Movie Rating Statistics ---")
print(f"Mean Rating: {rating_mean:.2f}")
print(f"Median Rating: {rating_median:.2f}")
print(f"Standard Deviation of Rating: {rating_std:.2f}")

# 2. Budget Statistics (Million $)
budget_mean = np.mean(df['Budget_Million'])
budget_median = np.median(df['Budget_Million'])
budget_std = np.std(df['Budget_Million'])

print("\n--- Movie Budget Statistics (Million $) ---")
print(f"Mean Budget: {budget_mean:.2f}")
print(f"Median Budget: {budget_median:.2f}")
print(f"Standard Deviation of Budget: {budget_std:.2f}")

# 3. Revenue Statistics (Million $)
revenue_mean = np.mean(df['Revenue_Million'])
revenue_median = np.median(df['Revenue_Million'])
revenue_std = np.std(df['Revenue_Million'])

print("\n--- Movie Revenue Statistics (Million $) ---")
print(f"Mean Revenue: {revenue_mean:.2f}")
print(f"Median Revenue: {revenue_median:.2f}")
print(f"Standard Deviation of Revenue: {revenue_std:.2f}")
```

```python
import pandas as pd

# Part 4: Grouped Analysis with Pandas

# Load the cleaned movie ratings dataset
df = pd.read_csv('cleaned_movie_ratings_data.csv')

# 1. Group by Genre and compute average ratings
genre_avg_rating = df.groupby('Genre')['Rating'].mean()
print("--- Average Rating by Genre ---")
print(genre_avg_rating)

# 2. Group by Director to analyze average rating per director
director_avg_rating = df.groupby('Director')['Rating'].mean()
print("\n--- Average Rating by Director ---")
print(director_avg_rating)

# 3. Group by Year to analyze rating trends over time
year_avg_rating = df.groupby('Year')['Rating'].mean()
print("\n--- Average Rating by Year ---")
print(year_avg_rating)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Part 5: Visualizations with Matplotlib

# Load the cleaned movie ratings dataset
df = pd.read_csv('cleaned_movie_ratings_data.csv')

# 1. Bar Plot: Average Rating by Genre and by Director
genre_avg = df.groupby('Genre')['Rating'].mean()
plt.figure(figsize=(10, 5))
genre_avg.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Average Rating by Genre')
plt.xlabel('Genre')
plt.ylabel('Average Rating')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

director_avg = df.groupby('Director')['Rating'].mean()
plt.figure(figsize=(12, 5))
director_avg.plot(kind='bar', color='salmon', edgecolor='black')
plt.title('Average Rating by Director')
plt.xlabel('Director')
plt.ylabel('Average Rating')
plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 2. Line Plot: Average Ratings Over Years
year_avg = df.groupby('Year')['Rating'].mean()
plt.figure(figsize=(10, 5))
year_avg.plot(kind='line', marker='o', color='green', linestyle='-', linewidth=2)
plt.title('Average Ratings Over Years')
plt.xlabel('Year')
plt.ylabel('Average Rating')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 3. Scatter Plot: Budget vs. Rating & Revenue vs. Rating
plt.figure(figsize=(10, 5))
plt.scatter(df['Budget_Million'], df['Rating'], color='purple', alpha=0.7, edgecolors='black')
plt.title('Budget vs. Rating')
plt.xlabel('Budget (Million $)')
plt.ylabel('Rating')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

plt.figure(figsize=(10, 5))
plt.scatter(df['Revenue_Million'], df['Rating'], color='teal', alpha=0.7, edgecolors='black')
plt.title('Revenue vs. Rating')
plt.xlabel('Revenue (Million $)')
plt.ylabel('Rating')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# 4. Horizontal Bar Plot: Top-Rated Movies
top_movies = df.sort_values(by='Rating', ascending=False).head(10)
plt.figure(figsize=(10, 6))
plt.barh(top_movies['Movie_Title'], top_movies['Rating'], color='orange', edgecolor='black')
plt.title('Top-Rated Movies')
plt.xlabel('Rating')
plt.ylabel('Movie Title')
plt.gca().invert_yaxis()  # Highest rated movie at the top
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
